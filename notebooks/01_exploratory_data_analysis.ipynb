# Objective

This notebook reviews the synthetic SEO content performance dataset and explains the main dataset qualities before modeling.

## Purpose

The notebook is the EDA and contract review for the synthetic dataset. It should establish the prediction-time feature window, show the label definition, and prepare the reader for the safe and leaky model comparison that follows later.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw/seo_content_performance.csv')
df.head()
```

# Data Exploration

The dataset is synthetic and pseudonymized. It contains features from content metadata, SEO metrics, user engagement, and historical ranking behavior.

The data contract in the repository expects columns such as `content_id`, `content_type`, `category`, `word_count`, `content_age_days`, `author_type`, `has_schema`, `organic_clicks`, `organic_impressions`, `ctr`, `average_position`, `keyword_count`, `ranking_keywords`, `backlinks`, `domain_authority`, `sessions`, `bounce_rate`, `avg_session_duration`, `conversions`, `clicks_7d`, `clicks_30d`, `impressions_7d`, `impressions_30d`, `position_7d`, `position_30d`, and the label `is_declining_label`.


```python
df.shape
df.dtypes
df.isna().sum()
df.duplicated().sum()
df.describe(include='all').T

# Print the expected data health summary clearly.
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
print(f"Missing values: {int(df.isna().sum().sum())}")
print(f"Duplicate records: {int(df.duplicated().sum())}")

# Example of the feature window that the contract expects.
print("Feature columns include content metadata, organic performance, engagement, and historical windows.")
```

# Target Distribution

The target is a binary label named `is_declining_label`. We inspect whether the class distribution is balanced enough for a baseline classification workflow.

The target distribution is a useful sanity check before modeling because a highly skewed target can make precision and recall interpretation more important than accuracy alone.


```python
sns.countplot(data=df, x='is_declining_label')
plt.title('Target distribution')
plt.show()

# Show class balance explicitly as a small table.
print(df['is_declining_label'].value_counts().sort_index())
```
