# Objective

This notebook reviews the synthetic SEO content performance dataset and explains the main dataset qualities before modeling.

# Load Data

We load the synthetic dataset and inspect the shape, missing values, and basic field types.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('../data/raw/seo_content_performance.csv')
df.head()
```

# Data Exploration

The dataset is synthetic and pseudonymized. It contains features from content metadata, SEO metrics, user engagement, and historical ranking behavior.

```python
df.shape
df.dtypes
df.isna().sum()
df.duplicated().sum()
df.describe(include='all').T
```

# Target Distribution

The target is a binary label named `is_declining_label`. We will inspect whether the class distribution is balanced enough for a baseline classification workflow.

```python
sns.countplot(data=df, x='is_declining_label')
plt.title('Target distribution')
plt.show()
```

# Feature Relationships

The notebook should identify suspicious relationships such as historical ranking metrics or future metrics that map strongly to the target.

```python
corr = df[[
    'ctr', 'average_position', 'organic_clicks', 'organic_impressions', 'bounce_rate',
    'clicks_7d', 'clicks_30d', 'impressions_7d', 'impressions_30d', 'is_declining_label'
].corr()
sns.heatmap(corr, annot=False)
plt.title('Feature correlation heatmap')
plt.show()
```

# Findings

The EDA focuses on whether the dataset has real variability, whether suspicious future columns exist, and whether the target can be derived from future data as designed.
